4. Merge z pozostałymi kolumnami stałymi per SKU

```python
# Krok 1: Cena_realna na oryginalnych rekordach sprzedaży
df_fi_hard_21['Cena_realna'] = (
    df_fi_hard_21['Wartosc'] / df_fi_hard_21['IloscPlus']
).round(2)

# Gdzie IloscPlus=0 zastąp CenaPoRab
mask_zero = df_fi_hard_21['IloscPlus'] == 0
df_fi_hard_21.loc[mask_zero, 'Cena_realna'] = df_fi_hard_21.loc[mask_zero, 'CenaPoRab']

print(f"NaN w Cena_realna: {df_fi_hard_21['Cena_realna'].isna().sum()}")
print(df_fi_hard_21['Cena_realna'].describe())
```

```python
# Krok 2: Daty start/koniec per SKU
data_max_global = df_fi_hard_21['Data'].max()

daty_sku = (
    df_fi_hard_21
    .groupby('TowId')['Data']
    .agg(DataStart='min', DataKoniec='max')
    .reset_index()
)

# DataKoniec + 14 dni (nie więcej niż globalne max)
daty_sku['DataKoniec'] = (
    (daty_sku['DataKoniec'] + pd.Timedelta(days=14))
    .clip(upper=data_max_global)
)

# Budowa kalendarza per SKU
kalendarze = []
for _, row in daty_sku.iterrows():
    dni = pd.date_range(start=row['DataStart'], end=row['DataKoniec'], freq='D')
    kalendarze.append(pd.DataFrame({'TowId': row['TowId'], 'Data': dni}))

kalendarz_full = pd.concat(kalendarze, ignore_index=True)
print(f"Rozmiar kalendarza: {len(kalendarz_full):,} rekordów")
print(f"Unikalnych SKU: {kalendarz_full['TowId'].nunique()}")
```

```python
# Krok 3 — agregacja dzienna sprzedaży per TowId i merge z kalendarzem: 

df_dzienne = (
    df_fi_hard_21
    .groupby(['TowId', 'Data'])
    .agg(
        Sprzedaz_szt=('IloscPlus', 'sum'),
        Wartosc=('Wartosc', 'sum'),
        Liczba_trans=('DokId', 'nunique'),
        CenaDet=('CenaDet', 'last'),
        CenaPoRab=('CenaPoRab', 'last'),
        Cena_realna=('Cena_realna', 'last'),
    )
    .reset_index()
)

# Merge z kalendarzem
kalendarz_full = kalendarz_full.merge(df_dzienne, on=['TowId', 'Data'], how='left')

# Wypełnij zera
kalendarz_full[['Sprzedaz_szt', 'Wartosc', 'Liczba_trans']] = (
    kalendarz_full[['Sprzedaz_szt', 'Wartosc', 'Liczba_trans']].fillna(0)
)

# ffill cen per SKU
kalendarz_full[['CenaDet', 'CenaPoRab', 'Cena_realna']] = (
    kalendarz_full
    .groupby('TowId')[['CenaDet', 'CenaPoRab', 'Cena_realna']]
    .ffill()
)

print(f"Shape: {kalendarz_full.shape}")
print(f"NaN w CenaDet: {kalendarz_full['CenaDet'].isna().sum()}")
print(f"NaN w Cena_realna: {kalendarz_full['Cena_realna'].isna().sum()}")
```

```python
# Krok 4: merge kolumn stałych per SKU z df_fi_hard:

sku_info = (
    df_fi_hard[['TowId', 'AsId', 'NazwaTow', 'NazwaAsort', 'CzyWazone']]
    .drop_duplicates('TowId')
)

kalendarz_full = kalendarz_full.merge(sku_info, on='TowId', how='left')

print(f"Shape: {kalendarz_full.shape}")
print(f"NaN w NazwaTow: {kalendarz_full['NazwaTow'].isna().sum()}")
print(f"NaN w AsId: {kalendarz_full['AsId'].isna().sum()}")
```
