```python
import pandas as pd
from utils.tools import hash_danych_bezpieczny
```

```python
RAW_PATH = "dane/raw/"
INTERIM_PATH = "dane/interim/"
```

```python
# Sumy kontrolne wejściowe - różne kolumny sortowania per plik
hashe_wejsciowe = [
    ('asort', f"{RAW_PATH}asort.parquet", ['AsId']),
    ('pozdok', f"{RAW_PATH}pozdok.parquet", ['TowId', 'DokId']),
    ('dok', f"{RAW_PATH}dok.parquet", ['DokId']),
    ('towar', f"{RAW_PATH}towar.parquet", ['TowId'])
]

for nazwa, plik, kolumny in hashe_wejsciowe:
    h = hash_danych_bezpieczny(plik, kolumny_sortowania=kolumny)
    print(f"Hash {nazwa}: {h}")
```

```python
df_asort = pd.read_parquet(f"{RAW_PATH}asort.parquet")
df_pozd = pd.read_parquet(f"{RAW_PATH}pozdok.parquet")
df_dok = pd.read_parquet(f"{RAW_PATH}dok.parquet")
df_tow = pd.read_parquet(f"{RAW_PATH}towar.parquet")
```

```python
# df_tow
tow_cols = ['TowId', 'AsId', 'Nazwa', 'Kod', 'Opis1', 'Producent', 'Aktywny']
df_tow_selected = df_tow[tow_cols].copy()
df_tow_selected = df_tow_selected.rename(columns={'Nazwa': 'NazwaTow', 'Aktywny': 'AktywnyTow', 'Kod': 'EAN'})

# df_asort
asort_cols = ['AsId', 'Nazwa']
df_asort_selected = df_asort[asort_cols].copy()
df_asort_selected = df_asort_selected.rename(columns={'Nazwa': 'NazwaAsort'})

# df_dok
dok_cols = ['DokId', 'Data', 'KolejnyWDniu', 'NrDok', 'TypDok', 'Aktywny', 'Razem', 'DoZaplaty', 'Zaplacono']
df_dok_selected = df_dok[dok_cols].copy()
df_dok_selected = df_dok_selected.rename(columns={'Aktywny': 'AktywnyDok'})

# df_pozd
pozd_cols = ['DokId', 'Kolejnosc', 'NrPozycji', 'TowId', 'TypPoz', 'IloscPlus', 'IloscMinus', 'CenaPoRab', 'Wartosc']
df_pozd_selected = df_pozd[pozd_cols].copy()

print(f"df_tow_selected:   {df_tow_selected.shape}")
print(f"df_asort_selected: {df_asort_selected.shape}")
print(f"df_dok_selected:   {df_dok_selected.shape}")
print(f"df_pozd_selected:   {df_pozd_selected.shape}")
```

```python
df_tow_selected.to_parquet(f"{INTERIM_PATH}towar_selected.parquet", compression='zstd', index=False)
df_asort_selected.to_parquet(f"{INTERIM_PATH}asort_selected.parquet", compression='zstd', index=False)
df_dok_selected.to_parquet(f"{INTERIM_PATH}dok_selected.parquet", compression='zstd', index=False)
df_pozd_selected.to_parquet(f"{INTERIM_PATH}pozdok_selected.parquet", compression='zstd', index=False)
```

#### Parametry dla kompresji przy zapisie **parquet**:
- **snappy** najszybszy odczyt, słabsza kompresja - domyślny 
- **gzip** dobry balans
- **zstd** bardzo dobry balans (szybkość + kompresja)
- **brotli** najmniejszy plik, wolniejszy odczyt
